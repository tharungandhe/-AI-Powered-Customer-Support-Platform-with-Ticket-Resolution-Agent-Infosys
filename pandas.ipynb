```python
import pandas as pd

# Load the file and skip any corrupted rows that break the parser
df = pd.read_csv('customer_support_tickets.csv', on_bad_lines='skip')

# Grab the first 500 rows
df_500 = df.head(500)

# Display the clean table
display(df_500)
```

#STEP-1 Missing values

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
df.fillna(0,inplace=True)
df.head()
```

#Step 2 Duplicates

```python
df.duplicated().sum()   ##to see the how any duplicates 
```

```python
df.drop_duplicates(inplace=True)  #to  remove the duplicates
```

```python
df.head()
```

##Step 3 Data Types

```python
df.dtypes
```

## Step 4 Handle the outlier 

```python
df.columns = df.columns.astype(str).str.strip() # 1. Clean column names

df['Customer Age'] = pd.to_numeric(df['Customer Age'], errors='coerce')
print(df['Customer Age'].describe())

df = df[df['Customer Age'] < 100]

print(f"\nRemaining rows after filtering: {len(df)}")
```

##Step 5 String Operations


```python
df['Ticket Channel'] = df['Ticket Channel'].str.strip()

df['Ticket Channel'] = df['Ticket Channel'].str.lower()  ##  Convert all text to lowercase to ensure consistency (e.g., "Phone" becomes "phone")
print(df['Ticket Channel'].value_counts())
```

##step 6 Encoding the Categorical Variable

```python
df = pd.concat([df, pd.get_dummies(df['Ticket Priority'])], axis=1)

# Label encoding
df['Priority_encoded'] = LabelEncoder().fit_transform(df['Ticket Priority'])

print(df[['Ticket Priority', 'Priority_encoded']].head())
```
